#### DESCARGA DE ARCHIVOS

```python
# URL del dataset normalizado en formato ZIP de FAOSTAT
url_datos_normalizados = "https://bulks-faostat.fao.org/production/Emissions_Totals_E_All_Data_(Normalized).zip"

# Descargamos el ZIP
result = requests.get(url_datos_normalizados)

with open("datos_normalizados.zip", "wb") as archivo:
  archivo.write(result.content)

print("El archivo ZIP se descargó correctamente")

# Descomprimimos el archivo
with zipfile.ZipFile("datos_normalizados.zip", "r") as archivo_zip:
    archivo_zip.extractall("datos_normalizados")

print("El archivo ZIP se descomprimió correctamente")
```
